# Importing necessary libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Loading The Raw Data

```python
Raw_Data = open("../data/Raw_Data1.txt","r") 
#Opens the text file in readable format 
```

```python
content = Raw_Data.read() #reading the file 
Raw_list = content.split() # converts the sequences into a list for easier iteration
```

```python
DF = pd.DataFrame({
    'Serial_No': range(1, len(Raw_list)+1),
    'DNA_Sample': Raw_list,
})
#Creating a DataFrame as a base for the DNA Samples
```

# Writing detector block for each disease 

```python
def check(sequence, motif):
    max_count = 0 
    count = 0
    i = 0
    while i <= len(sequence) - len(motif): #consider the length of the pattern too 
        if sequence[i:i + len(motif)] == motif: #condition to match pattern(different for each disease 
            count += 1
            i += len(motif)
            max_count = max(max_count, count)
        else:
            count = 0
            i += 1
    return max_count 
```

```python
'''
each column for a disease is preset to none 
if the sample contains the disese the column says 'Found' 
the count of repetitions is given in the column next to it 
'''
```

### Code block for Huntington's Disease

```python
# 1.Huntington's Disease
DF['Huntington\'s_Disease'] = "Not Found" #column to hold boolean value 
```

```python
DF['HD_repetitions'] = np.nan  # Create empty column

for idx, row in DF.iterrows():
    sequence = row['DNA_Sample']
    max_count = check(sequence, motif='CAG')
    if max_count >= 36:
        DF.at[idx, 'HD_repetitions'] = max_count  # Pathogenic repeat count
        DF.at[idx, 'Huntington\'s_Disease'] = "Found" 
```
